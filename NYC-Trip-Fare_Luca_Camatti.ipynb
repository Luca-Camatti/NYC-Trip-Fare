**11) For each payment type, determine the interval when the overall ratio between the tip and the fare amounts is maximum**

```python
import pandas as pd

# Si carica il dataset
df = pd.read_csv("data.csv", low_memory=False)

# Ci si assicura che il campo datetime sia in formato datetime
df["tpep_pickup_datetime"] = pd.to_datetime(df["tpep_pickup_datetime"])

# Si crea la colonna pickup_time_interval (30 minuti)
def cluster_pickup_time(dt):
    hour = dt.hour
    minute = dt.minute
    if minute < 30:
        return f"{hour:02d}:00 - {hour:02d}:30"
    else:
        next_hour = (hour + 1) % 24
        return f"{hour:02d}:30 - {next_hour:02d}:00"

df["pickup_time_interval"] = df["tpep_pickup_datetime"].apply(cluster_pickup_time)

# Si filtrano le righe che contengono valori validi
df = df[(df["fare_amount"] > 0) & df["tip_amount"].notnull()]

# Si raggruppa per payment_type e intervalli di tempo e si calcola la somma delle tip e delle fare
grouped = df.groupby(["payment_type", "pickup_time_interval"]).agg(
    total_tip=("tip_amount", "sum"),
    total_fare=("fare_amount", "sum")
).reset_index()

# Si calcola il rapporto totale tip/fare per ogni gruppo
grouped["tip_fare_ratio"] = grouped["total_tip"] / grouped["total_fare"]

# Si trova l'intervallo con il rapporto massimo per ciascun payment_type
max_ratios = grouped.loc[grouped.groupby("payment_type")["tip_fare_ratio"].idxmax()]

# Si arrotonda per leggibilità
max_ratios['tip_fare_ratio_percent'] = (max_ratios['tip_fare_ratio'] * 100).round(2).astype(str) + '%'

# Si Visualizza il risultato
print(max_ratios[["payment_type", "pickup_time_interval", "tip_fare_ratio_percent"]])
```

**13) Build a new dataframe (called common) where, for each pickup location we keep all trips to the 5 most common destinations (i.e. each pickup location can have different common destinations).**

```python
import pandas as pd

# Si caricano i dati
df = pd.read_csv("data.csv", low_memory=False)

# Si crea una funzione per ottenere le 5 destinazioni più comuni da una pickup location
def top_5_destinations(df):
    return (
        df.groupby(["PULocationID", "DOLocationID"])
          .size()
          .reset_index(name="count")
    )

# Si calcola la frequenza per tutte le coppie pickup-dropoff
pair_counts = top_5_destinations(df)

# Si ordinano quest'ultime per count decrescente all'interno di ogni PULocationID
pair_counts = pair_counts.sort_values(["PULocationID", "count"], ascending=[True, False])

# Si prendono le prime 5 DOLocationID per ciascun PULocationID
top5_pairs = pair_counts.groupby("PULocationID").head(5)

# Si fa il merge con il dataset originale per ottenere solo i viaggi corrispondenti
common = df.merge(top5_pairs[["PULocationID", "DOLocationID"]], on=["PULocationID", "DOLocationID"])

# Si mostra il nuovo dataframe "common"
print("Nuovo DataFrame 'common' con le 5 destinazioni più comuni per ciascuna pickup location:")
print(common)
```

**15) Compute the difference of the average fare amount computed in the previous point with those computed at point 9**

```python
import pandas as pd

# Ci si assicura che il datetime sia in formato corretto sul df completo
df["tpep_pickup_datetime"] = pd.to_datetime(df["tpep_pickup_datetime"])

# Si definisce la funzione per creare gli intervalli di 30 minuti
def cluster_pickup_time(dt):
    hour = dt.hour
    minute = dt.minute
    if minute < 30:
        return f"{hour:02d}:00 - {hour:02d}:30"
    else:
        next_hour = (hour + 1) % 24
        return f"{hour:02d}:30 - {next_hour:02d}:00"

if "pickup_time_interval" not in df.columns:
    df["pickup_time_interval"] = df["tpep_pickup_datetime"].apply(cluster_pickup_time)

# Si filtrano solo le righe valide
filtered_full = df[df["fare_amount"] > 0]

# Si calcola la media del fare per payment_type e intervallo (dataset completo)
avg_fare_full = (
    filtered_full.groupby(["payment_type", "pickup_time_interval"])["fare_amount"]
    .mean()
    .reset_index()
)
avg_fare_full.rename(columns={"fare_amount": "avg_fare_full"}, inplace=True)

# --- Punto 14: media del fare_amount per payment_type e intervallo sul dataframe common ---

# Ci si assicura che il datetime sia in formato corretto sul dataframe common
common["tpep_pickup_datetime"] = pd.to_datetime(common["tpep_pickup_datetime"])
if "pickup_time_interval" not in common.columns:
    common["pickup_time_interval"] = common["tpep_pickup_datetime"].apply(cluster_pickup_time)

# Si filtra solo le righe valide
filtered_common = common[common["fare_amount"] > 0]

# Si calcola la media fare per payment_type e intervallo (solo viaggi comuni)
avg_fare_common = (
    filtered_common.groupby(["payment_type", "pickup_time_interval"])["fare_amount"]
    .mean()
    .reset_index()
)
avg_fare_common.rename(columns={"fare_amount": "avg_fare_common"}, inplace=True)

# --- Calcolo della differenza tra avg_fare_common e avg_fare_full ---

# Si esegue la merge tra i due dataframe sulle colonne payment_type e pickup_time_interval
fare_comparison = pd.merge(
    avg_fare_common,
    avg_fare_full,
    on=["payment_type", "pickup_time_interval"],
    how="inner"
)

# Si effettua il calcolo della differenza (common - full)
fare_comparison["fare_diff"] = fare_comparison["avg_fare_common"] - fare_comparison["avg_fare_full"]

# Si arrotonda per chiarezza
fare_comparison["avg_fare_common"] = fare_comparison["avg_fare_common"].round(2)
fare_comparison["avg_fare_full"] = fare_comparison["avg_fare_full"].round(2)
fare_comparison["fare_diff"] = fare_comparison["fare_diff"].round(2)

# Si mostra il risultato
print("Differenza media fare_amount (common - full) per payment_type e intervallo:")
print(fare_comparison)
```

**16) Compute the ratio between the differences computed in the previous point and those computed in point 9. Note: you have to compute a ratio for each pair (payment type, interval).**

```python
# Calcolo del rapporto tra la differenza e la media fare completa (punto 9)

# Si evitano divisioni per zero sostituendo 0 con NaN temporaneamente
fare_comparison["fare_ratio"] = fare_comparison.apply(
    lambda row: row["fare_diff"] / row["avg_fare_full"] if row["avg_fare_full"] != 0 else None,
    axis=1
)

# Si arrotonda a 4 decimali per chiarezza
fare_comparison["fare_ratio"] = fare_comparison["fare_ratio"].round(4)

# Si visualizza il risultato
print("Rapporto tra la differenza media (common - full) e la media completa del fare_amount:")
print(fare_comparison[["payment_type", "pickup_time_interval", "fare_diff", "avg_fare_full", "fare_ratio"]])
```

**17) Build chains of trips. Two trips are consecutive in a chain if (a) they have the same VendorID, (b) the pickup location of the second trip is also the dropoff location of the first trip, (c) the pickup time of the second trip is after the dropoff time of the first trip, and (d) the pickup time of the second trip is at most 2 minutes later than the dropoff time of the first trip.**

```python
import pandas as pd

# Assicuriamoci che i datetime siano corretti
df["tpep_pickup_datetime"] = pd.to_datetime(df["tpep_pickup_datetime"])
df["tpep_dropoff_datetime"] = pd.to_datetime(df["tpep_dropoff_datetime"])

# Ordiniamo per VendorID e pickup/dropoff datetime
df = df.sort_values(["VendorID", "tpep_pickup_datetime"]).reset_index(drop=True)

# Copie separate per il merge
df_current = df.copy()
df_next = df.copy()

# Rinominiamo le colonne di df_next per distinguerle
df_next = df_next.rename(columns={
    "tpep_pickup_datetime": "next_pickup_datetime",
    "PULocationID": "next_PULocationID",
    "DOLocationID": "next_DOLocationID"
})

# Ordiniamo per merge_asof
df_current = df_current.sort_values(["VendorID", "tpep_dropoff_datetime"])
df_next = df_next.sort_values(["VendorID", "next_pickup_datetime"])

# Applichiamo merge_asof con tolleranza di 2 minuti
merged = pd.merge_asof(
    df_current,
    df_next,
    left_on="tpep_dropoff_datetime",
    right_on="next_pickup_datetime",
    by="VendorID",
    direction="forward",
    tolerance=pd.Timedelta("2min")
)

# Filtriamo solo i casi in cui il dropoff della prima corsa coincide col pickup della successiva
chains = merged[merged["DOLocationID"] == merged["next_PULocationID"]]

# Visualizziamo le catene valide
print("Catene trovate:")
print(chains[[
    "VendorID",
    "tpep_pickup_datetime", "PULocationID", "DOLocationID", "tpep_dropoff_datetime",
    "next_pickup_datetime", "next_PULocationID", "next_DOLocationID"
]])
```
