**11) For each payment type, determine the interval when the overall ratio between the tip and the fare amounts is maximum**

```python
import pandas as pd

df = pd.read_csv("data.csv", low_memory=False)

df["tpep_pickup_datetime"] = pd.to_datetime(df["tpep_pickup_datetime"])
df["payment_type"] = df["payment_type"].astype("Int64")

# The 30-minute intervals are calculated
def cluster_pickup_time(dt):
    hour = dt.hour
    minute = dt.minute
    if minute < 30:
        return f"{hour:02d}:00 - {hour:02d}:30"
    else:
        next_hour = (hour + 1) % 24
        return f"{hour:02d}:30 - {next_hour:02d}:00"

df["pickup_time_interval"] = df["tpep_pickup_datetime"].apply(cluster_pickup_time)

# Only rows containing fare_amount and tip_amount greater than zero are kept
df = df[(df["fare_amount"] > 0) & df["tip_amount"].notnull()]

values = df.groupby(["payment_type", "pickup_time_interval"]).agg(
    total_tip=("tip_amount", "sum"),
    total_fare=("fare_amount", "sum")
).reset_index()

# The total tips/fares ratio for each group is calculated
values["tip_fare_ratio"] = values["total_tip"] / values["total_fare"]

# The interval with the maximum ratio for each payment_type is found
max_ratios = values.loc[values.groupby("payment_type")["tip_fare_ratio"].idxmax()]

max_ratios['tip_fare_ratio_percent'] = (max_ratios['tip_fare_ratio'] * 100).round(2).astype(str) + '%'

print(max_ratios[["payment_type", "pickup_time_interval", "tip_fare_ratio_percent"]])
```

**13) Build a new dataframe (called common) where, for each pickup location we keep all trips to the 5 most common destinations (i.e. each pickup location can have different common destinations).**

```python
import pandas as pd

df = pd.read_csv("data.csv", low_memory=False)
df["VendorID"] = df["VendorID"].astype("Int64")
df["passenger_count"] = df["passenger_count"].astype("Int64")
df["payment_type"] = df["payment_type"].astype("Int64")

# A function is created to get the 5 most common destinations from each pickup location
def top_5_destinations(df):
    return (
        df.groupby(["PULocationID", "DOLocationID"])
          .size()
          .reset_index(name="count")
    )

# The function is applied
pair_counts = top_5_destinations(df)

# The first 5 DOLocationIDs are taken for each PULocationID
pair_counts = pair_counts.sort_values(["PULocationID", "count"], ascending=[True, False])
top5_pairs = pair_counts.groupby("PULocationID").head(5)

# The rows are merged with the original dataset to obtain only the corresponding trips
common = df.merge(top5_pairs[["PULocationID", "DOLocationID"]], on=["PULocationID", "DOLocationID"])

print(common)
```

**15) Compute the difference of the average fare amount computed in the previous point with those computed at point 9**

```python
import pandas as pd

df["tpep_pickup_datetime"] = pd.to_datetime(df["tpep_pickup_datetime"])

# The pickup_time_interval column is created
def cluster_pickup_time(dt):
    hour = dt.hour
    minute = dt.minute
    if minute < 30:
        return f"{hour:02d}:00 - {hour:02d}:30"
    else:
        next_hour = (hour + 1) % 24
        return f"{hour:02d}:30 - {next_hour:02d}:00"

if "pickup_time_interval" not in df.columns:
    df["pickup_time_interval"] = df["tpep_pickup_datetime"].apply(cluster_pickup_time)

# Only rows containing fare amounts greater than zero are kept
filtered_df = df[df["fare_amount"] > 0]

# The mean of fare_amount by payment_type and interval is calculated (original dataframe df)
avg_fare_df = (filtered_df.groupby(["payment_type", "pickup_time_interval"])
    ["fare_amount"].mean().reset_index())

avg_fare_df.rename(columns={"fare_amount": "avg_fare_df"}, inplace=True)

# ------ 14) On the common dataframe, for each payment type and each interval, determine the average fare amount ------

# The datetime pickup column is converted to datetime format (common dataframe)
common["tpep_pickup_datetime"] = pd.to_datetime(common["tpep_pickup_datetime"])
if "pickup_time_interval" not in common.columns:
    common["pickup_time_interval"] = common["tpep_pickup_datetime"].apply(cluster_pickup_time)

# Only rows containing fare amounts greater than zero are kept (common dataframe)
filtered_common = common[common["fare_amount"] > 0]

# The mean of fare_amount by payment_type and interval is calculated (common journeys only)
avg_fare_common = (filtered_common.groupby(["payment_type", "pickup_time_interval"])
    ["fare_amount"].mean().reset_index())

avg_fare_common.rename(columns={"fare_amount": "avg_fare_common"}, inplace=True)

# --- Calculating the delta between avg_fare_common and avg_fare_df ---

# The two dataframes are merged on the payment_type and pickup_time_interval columns
fare_comparison = pd.merge(
    avg_fare_common,
    avg_fare_df,
    on=["payment_type", "pickup_time_interval"],
    how="inner"
)

# Calculating the delta between average fares (common - df)
fare_comparison["fare_diff"] = fare_comparison["avg_fare_common"] - fare_comparison["avg_fare_df"]

fare_comparison["avg_fare_common"] = fare_comparison["avg_fare_common"].round(2)
fare_comparison["avg_fare_df"] = fare_comparison["avg_fare_df"].round(2)
fare_comparison["fare_diff"] = fare_comparison["fare_diff"].round(2)

print(fare_comparison)
```

**17) Build chains of trips. Two trips are consecutive in a chain if (a) they have the same VendorID, (b) the pickup location of the second trip is also the dropoff location of the first trip, (c) the pickup time of the second trip is after the dropoff time of the first trip, and (d) the pickup time of the second trip is at most 2 minutes later than the dropoff time of the first trip.**

```python
import pandas as pd

df["tpep_pickup_datetime"] = pd.to_datetime(df["tpep_pickup_datetime"], errors="coerce")
df["tpep_dropoff_datetime"] = pd.to_datetime(df["tpep_dropoff_datetime"], errors="coerce")
df["VendorID"] = df["VendorID"].astype(int)

# Rows with missing data are removed
df = df.dropna(subset=["VendorID", "tpep_pickup_datetime", "tpep_dropoff_datetime", "PULocationID", "DOLocationID"])

# Df_next is created by renaming columns for the next trip
df_next = df.rename(columns={
    "tpep_pickup_datetime": "next_pickup_datetime",
    "PULocationID": "next_PULocationID",
    "DOLocationID": "next_DOLocationID"
})

df_current = df.sort_values(by=["VendorID", "tpep_dropoff_datetime"]).reset_index(drop=True)
df_next = df_next.sort_values(by=["VendorID", "next_pickup_datetime"]).reset_index(drop=True)

# Visual Debug: Displays the first rows of both dataframes to check sorting
print("=== df_current (sorted by VendorID, tpep_dropoff_datetime) ===")
print(df_current.head(10))

print("\n=== df_next (sorted by VendorID, next_pickup_datetime) ===")
print(df_next.head(10))

# Assert control for security
assert df_current["VendorID"].is_monotonic_increasing 
and df_current["tpep_dropoff_datetime"].is_monotonic_increasing, "df_current not sorted"
assert df_next["VendorID"].is_monotonic_increasing 
and df_next["next_pickup_datetime"].is_monotonic_increasing, "df_next not sorted"

# Execute merge_asof with 2 minutes tolerance
merged = pd.merge_asof(
    df_current,
    df_next,
    left_on="tpep_dropoff_datetime",
    right_on="next_pickup_datetime",
    by="VendorID",
    direction="forward",
    tolerance=pd.Timedelta("2min")
)

# Only valid chains are kept (dropoff = next pickup)
chains = merged[merged["DOLocationID"] == merged["next_PULocationID"]]

print(chains[["VendorID", 
              "tpep_pickup_datetime", 
              "PULocationID", 
              "DOLocationID", 
              "tpep_dropoff_datetime",
              "next_pickup_datetime", 
              "next_PULocationID", 
              "next_DOLocationID"
]])
```
